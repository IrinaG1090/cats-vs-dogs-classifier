```python
import os
import shutil
import random
from pathlib import Path

# Пути
base_dir = Path('D:/Projects/image-classification/data')
source_cats = base_dir / 'PetImages/Cat'
source_dogs = base_dir / 'PetImages/Dog'

train_cats = base_dir / 'train/cats'
train_dogs = base_dir / 'train/dogs'
val_cats = base_dir / 'validation/cats'
val_dogs = base_dir / 'validation/dogs'

# Создаём папки, если их нет
for folder in [train_cats, train_dogs, val_cats, val_dogs]:
    folder.mkdir(parents=True, exist_ok=True)

# Функция для разделения
def split_data(source, train_dest, val_dest, split_ratio=0.8):
    files = os.listdir(source)
    random.shuffle(files)
    
    split_point = int(len(files) * split_ratio)
    train_files = files[:split_point]
    val_files = files[split_point:]
    
    for f in train_files:
        src = source / f
        if src.is_file():  # проверяем, что это файл
            shutil.copy(src, train_dest / f)
    
    for f in val_files:
        src = source / f
        if src.is_file():
            shutil.copy(src, val_dest / f)
    
    print(f"Скопировано: {len(train_files)} в train, {len(val_files)} в validation")

# Разделяем кошек и собак
print("Копируем кошек...")
split_data(source_cats, train_cats, val_cats)

print("Копируем собак...")
split_data(source_dogs, train_dogs, val_dogs)

print("✅ Готово! Все фото скопированы.")
```
